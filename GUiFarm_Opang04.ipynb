```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- สร้าง UI Widgets ---
style = {'description_width': 'initial'}

header = widgets.HTML("<h2>โปรแกรมกันใช้ตังค์เกินงบที่กำหนด (Real-time)</h2>")

# 1. รับชื่อผู้ใช้ (Username) เป็นอักษร (str)
username_input = widgets.Text(description="ชื่อบัญชี (Username):", style=style)

# 2. รับจำนวนเงินที่จำกัดการใช้ (Limit) เป็นตัวเลข (int)
limit_input = widgets.IntText(description="จำนวนเงินที่จำกัด (Limit):", style=style)

# 3. รับจำนวนวันประหยัดเงินที่กำหนด (Day) เป็นตัวเลข (int)
day_input = widgets.IntText(description="จะประหยัดกี่วันดี (Day):", style=style)

# *เพิ่ม: ยอดเงินที่ใช้ไป (Spending) เพื่อให้ Logic ในภาพทำงานได้
spending_input = widgets.IntText(description="ยอดเงินที่ใช้ไปจริง (Spending):", style=style)

button = widgets.Button(description="ตรวจสอบสถานะ", button_style='primary')
output = widgets.Output()

def on_button_clicked(b):
    with output:
        clear_output()

        # ดึงค่าจาก Widgets
        username = username_input.value
        limit = limit_input.value
        day = day_input.value
        spending = spending_input.value

        print(f"สวัสดีครับคุณ: {username} (เป้าหมาย {day} วัน)")
        print("-" * 30)

        # พื้นที่แสดงผลลัพธ์ตาม Logic ในภาพ
        # if spending > limit:
        if spending > limit:
            print("คุณใช้เงินเกินกำหนด!!!!") #
        else:
            # คำนวณยอดคงเหลือ (ในภาพใช้ตัวแปร spending ใน f-string แต่ความหมายคือยอดเหลือ)
            balance = limit - spending
            print(f"ยอดคงเหลือในบัญชีของคุณคือ : {balance} บาท") #

button.on_click(on_button_clicked)

# --- จัดวางเลย์เอาต์และแสดงผล ---
ui = widgets.VBox([header, username_input, limit_input, day_input, spending_input, button, output])
display(ui)
```
