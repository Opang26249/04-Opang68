```python
import tkinter as tk
from tkinter import messagebox, font

def check_budget():
    try:
        user = entry_user.get()
        total_limit = float(entry_limit.get())
        total_days = int(entry_days.get())
        spent_today = float(entry_spent.get())

        # Logic การคำนวณ
        daily_budget = total_limit / total_days
        balance = total_limit - spent_today

        # สรุปข้อความสำหรับ Pop-up
        title = "📊 สรุปผลการใช้จ่าย"
        summary = f"คุณ {user}\n"
        summary += f"━━━━━━━━━━━━━━\n"
        summary += f"💰 ใช้จ่ายวันนี้: {spent_today:,.2f} บาท\n"
        summary += f"คงเหลือในบัญชี: {balance:,.2f} บาท\n"
        summary += f"━━━━━━━━━━━━━━\n"

        if spent_today > daily_budget:
            summary += "⚠️ สถานะ: คุณใช้เงินเกินกำหนดวันนี้!"
            messagebox.showwarning(title, summary)
        else:
            summary += "✅ สถานะ: คุณทำได้ดีมาก! ไม่เกินงบ"
            messagebox.showinfo(title, summary)

    except ValueError:
        messagebox.showerror("ข้อผิดพลาด", "กรุณากรอกเฉพาะตัวเลขในช่องจำนวนเงินและวัน")

# --- สร้างหน้าต่างหลัก ---
root = tk.Tk()
root.title("Budget Daily Tracker")
root.geometry("380x580")
root.configure(bg="#F8F9FA") # สีพื้นหลังเทาอ่อนๆ สบายตา

# กำหนดฟอนต์
title_font = ("Helvetica", 16, "bold")
label_font = ("Helvetica", 10)
entry_font = ("Helvetica", 12)

# ส่วน Header (แถบสีด้านบน)
header_frame = tk.Frame(root, bg="#6C63FF", height=80)
header_frame.pack(fill="x")
header_label = tk.Label(header_frame, text="My Budget Control", font=title_font, fg="white", bg="#6C63FF")
header_label.pack(pady=25)

# Container สำหรับเนื้อหา
content_frame = tk.Frame(root, bg="#F8F9FA", padx=30)
content_frame.pack(fill="both", expand=True, pady=20)

def create_input(label_text, entry_var_name, is_highlight=False):
    tk.Label(content_frame, text=label_text, font=label_font, bg="#F8F9FA", fg="#666").pack(anchor="w", pady=(10, 2))
    entry = tk.Entry(content_frame, font=entry_font, bd=0, highlightthickness=1,
                     highlightbackground="#DDD", highlightcolor="#6C63FF")
    if is_highlight:
        entry.config(highlightbackground="#6C63FF", highlightthickness=2)
    entry.pack(fill="x", ipady=8)
    return entry

# สร้างช่องกรอกข้อมูล
entry_user = create_input("ชื่อบัญชี (Username)", "user")
entry_limit = create_input("งบประมาณทั้งหมด (Limit)", "limit")
entry_days = create_input("จำนวนวันที่วางแผน (Day)", "days")
entry_spent = create_input("ยอดเงินที่ใช้จริงวันนี้ (Spending Today)", "spent", is_highlight=True)

# ปุ่มกดแบบมนๆ (Simulated Rounded Button)
btn_check = tk.Button(content_frame, text="ตรวจสอบยอดเงิน", command=check_budget,
                      bg="#6C63FF", fg="white", font=("Helvetica", 12, "bold"),
                      bd=0, cursor="hand2", activebackground="#5750D1", activeforeground="white")
btn_check.pack(fill="x", pady=30, ipady=10)

# ส่วนท้าย
footer_label = tk.Label(root, text="Designed for Financial Discipline", font=("Helvetica", 8), bg="#F8F9FA", fg="#AAA")
footer_label.pack(side="bottom", pady=10)

root.mainloop()
```
